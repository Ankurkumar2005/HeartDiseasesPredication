```python
def ploting(var,num):
    plt.subplot(2,2,num)
    sns.histplot(df[var].dropna(), kde=True, bins=20)

ploting('Age',1)
ploting('RestingBP',2) 
ploting('Cholesterol',3) 
ploting('MaxHR',4) 

plt.tight_layout()
```

```python
df['Cholesterol'].value_counts()
```

```python
ch_mean = df.loc[df['Cholesterol'] != 0, 'Cholesterol'].mean()
```

```python
ch_mean
```

```python
df['Cholesterol'] = df['Cholesterol'].replace(0,ch_mean)
df['Cholesterol'] = df['Cholesterol'].round(2)
```

```python
rs_mean = df.loc[df['RestingBP'] != 0, 'RestingBP'].mean()
df['RestingBP'] = df['RestingBP'].replace(0,rs_mean)
df['RestingBP'] = df['RestingBP'].round(2)
```

```python
def ploting(var,num):
    plt.subplot(2,2,num)
    sns.histplot(df[var].dropna(), kde=True, bins=20)

ploting('Age',1)
ploting('RestingBP',2) 
ploting('Cholesterol',3) 
ploting('MaxHR',4) 

plt.tight_layout()
```
